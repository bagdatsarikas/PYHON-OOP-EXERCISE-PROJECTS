Every smartphone is a phone. A phone can make calls, and a smartphone can make calls and take photos.

Your task:

Create a parent class phone:

__init__(self, br, nb) that saves self.brand and self.number
a method call(self, other) that prints: <brand> is calling <other>

Create a child class smartphone that inherits from phone:
no __init__
a new method take_photo(self) that prints: <brand> took a photo

```python
class phone:
    
    def __init__(self, brand, number):
        self.number = number
        self.brand = brand
        
    def call(self, person):
        print(self.brand, "is calling", person)
        
class smartphone(phone):
    
    def take_photo(self):
        print(self.brand, "takes a photo")

obj = smartphone("iphone", "344-55")
print(obj.call("mony"))
print(obj.take_photo())

obj2 = phone("samsung","455-55")
print(obj2.brand)  
print(obj2.number)
```

Story: Every pet has a name and an age. A cat also has a color, so the child needs to save something new. That means the child needs its own __init__.

Your task:

Create a parent class pet:
__init__(self, nm, ag) that saves self.name and self.age
a method info(self) that prints: <name> is <age> years old
    
Create a child class cat that inherits from pet:
    
its own __init__(self, nm, ag, cl):
use super().__init__ to save the name and age
then save self.color
a new method show_color(self) that prints: <name> is <color>

```python
class pet():
    
    def __init__(self, name, age):
        self.name = name 
        self.age = age 
        
    def info(self):
        print(self.name, "is", self.age, " years old")
        
class cat(pet):
    
    def __init__(self, name, age, color):
        super().__init__(name, age)
        self.color = color
        
    def showcolor(self):
        print("the color of ", self.name , "is", self.color)
        
c1 = cat("bar", 46, "black")
c2 = cat("sar", 45, "grey")

print(c1.info())
print(c1.showcolor())
print(c1.name)
print(c2.color)
```

Goal: use super().__init__() instead of copying lines.

Create a parent class vehicle:

__init__(self, br, sp) that saves self.brand and self.speed
display(self) that prints Brand: <brand>, Speed: <speed>
    
Create a child class car(vehicle):
    
__init__(self, br, sp, dr) that uses super().__init__ for brand and speed, then saves self.doors
override display(self) so it calls the parent’s display with super(), then prints Doors: <doors>

```python
class vehicle:                          # PARENT
    def __init__(self, br, sp):
        self.brand = br
        self.speed = sp

    def display(self):
        print("Brand:", self.brand, ", Speed:", self.speed)


class car(vehicle):                     
    def __init__(self, br, sp, dr):
        super().__init__(br, sp)        
        self.doors = dr                 

    def display(self):                  
        super().display()               # 1) run the parent's display first
        print("Doors:", self.doors)     # 2) then add the child's extra line

        
```

```python
ob = car("toyota", 180, 5)
print(ob)
```

### Problem: Make an iterator class Countdown that gives 5, 4, 3, 2, 1.

```python
class countdown:
    
    def __init__(self):
        self.num = 5 
        
    def __iter__(self):
        return self 
    
    def __next__(self):
        if self.num >= 1:
            value = self.num
            self.num = self.num - 1
            return value
        else : 
            raise StopIteration
            
for i in countdown():
    print(i)
```

### Write the same Countdown (5, 4, 3, 2, 1), but as a generator.

```python
def countdown():
    n = 5 
    while n >= 1:
        yield n 
        n = n - 1
        
for i in countdown():
    print(i)
```

### Your own iteratior : Write a class Squares that gives the squares of 1 to 5: 1, 4, 9, 16, 25.

```python

class squares:
    
    def __init__(self):
        self.num = 1 
        
    def __iter__(self):
        return self 
    
    def __next__(self):
        
        if self.num <= 5:
            val = self.num * self.num 
            self.num = self.num + 1 
            return val
        
        else:
            raise StopIteration
            
for i in squares():
    print(i)
```

```python
def squares():
    n = 1 
    while n <=5:
        yield n*n
        n = n + 1
for i in squares():
    print(i)
```
